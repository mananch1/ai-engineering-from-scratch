```python
def print_debug(name,tensor):
    print(f'name = {name} tensor_shape = {tensor.shape} tensor_dtype = {tensor.dtpye}')
    print(f'device = {tensor.device}')
    print(f'max = {tensor.max().item():.3f} min = {tensor.min().item():.3f}')
    print(f'mean = {tensor.mean().item():.5f}')
    print(f'has NaN = {tensor.isnan().any().item()}')
```
