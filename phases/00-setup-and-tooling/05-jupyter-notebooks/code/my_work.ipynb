```python
a = [2,7,3,3,9]

def is_sorted(a):
    m = 0
    for x in a :
        if(x < m):
            return False
        m = max(m,x)
    return True
```

```python
def smooth_array(B, X):
    S = 0
    B = B[:]

    for i in range(len(B)):
        if B[i] >= X:
            S += B[i] - X
            B[i] = X
        else:
            take = min(S, X - B[i])
            B[i] += take
            S -= take

    return B


# Example usage:

print(smooth_array(a, 2))  # [2, 2, 2, 2, 2]
print(smooth_array(a, 4))  # [4, 4, 3, 4, 2]
```

```python
#a = [1,2,10,3,4,20]
a = [2,4,4,7]
for x in range(100):
    if(is_sorted(smooth_array(a,x))):
        print(x)
```
